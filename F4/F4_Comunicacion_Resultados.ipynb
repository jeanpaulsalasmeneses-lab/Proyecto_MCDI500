## 3. Validación de los conjuntos de datos e indicadores analíticos

### 3.1. Propósito de la validación

Antes de desarrollar las visualizaciones finales, es necesario comprobar que los resultados recuperados del núcleo algorítmico presentan condiciones básicas de integridad y consistencia.

La validación se realiza sobre el archivo `F3_701_Panel_Integrado_Comunal.csv`, que consolida los indicadores correspondientes a las 15 comunas priorizadas.

El objetivo consiste en verificar que la estructura del panel sea coherente con el alcance del proyecto y que las tasas de frecuencia documental almacenadas correspondan a los conteos, las poblaciones de referencia y el período de observación establecido.

### 3.2. Metodología de validación

La validación inicial contempla cuatro procedimientos complementarios.

**Verificación de cardinalidad.** Se comprueba que el conjunto de datos contenga exactamente 15 registros, correspondientes a las comunas seleccionadas.

**Verificación de unicidad territorial.** Se identifican posibles duplicados en `codigo_comuna`, considerando que el panel debe contener un registro por comuna.

**Verificación de integridad.** Se comprueba la ausencia de valores faltantes en las variables fundamentales para la comparación territorial: código, nombre de comuna, población, cantidad de informes no anulados y frecuencia normalizada.

**Verificación de coherencia de la frecuencia.** Se recalcula la tasa documental utilizando los conteos de informes, la población del Censo 2024 y el período de observación de 33 meses. Posteriormente, se compara el resultado obtenido con la tasa almacenada en el panel integrado.

### 3.3. Verificación de la frecuencia documental normalizada

La frecuencia documental normalizada se calcula mediante la siguiente expresión:

\[
F_i=\frac{N_i}{P_i}\times100.000\times\frac{12}{33}
\]

Donde:

- \(F_i\): frecuencia documental normalizada de la comuna \(i\), expresada por 100.000 habitantes-año.
- \(N_i\): cantidad de informes documentales no anulados asociados con la comuna \(i\).
- \(P_i\): población de referencia de la comuna \(i\), según el Censo 2024.
- \(33\): período de observación expresado en meses.
- \(12/33\): factor utilizado para expresar la frecuencia en términos anuales.

La comparación se realiza mediante `numpy.isclose()`, utilizando una tolerancia absoluta de 0,011 y una tolerancia relativa igual a cero.

Esta tolerancia permite considerar diferencias menores asociadas con la representación decimal de las tasas almacenadas.

La verificación permite comprobar la coherencia aritmética entre las variables del panel, sin modificar los procedimientos implementados durante F3.

```python
# ============================================================
# MCDI500 - GRUPO 7
# FASE 4 - APARTADO 3
# VALIDACIÓN INICIAL DEL PANEL INTEGRADO
# ============================================================

import pandas as pd
import numpy as np

print("=" * 65)
print("FASE 4 - VALIDACIÓN DEL PANEL INTEGRADO")
print("=" * 65)

# 1. Trabajar con una copia para no modificar F3
datos = panel_f3.copy()

# 2. Verificar cantidad de comunas
print("\n1. CANTIDAD DE COMUNAS")

print("Registros encontrados:", len(datos))

if len(datos) == 15:
    print("RESULTADO: CORRECTO")
else:
    print("RESULTADO: REVISAR")


# 3. Verificar comunas duplicadas
print("\n2. COMUNAS DUPLICADAS")

duplicados = datos["codigo_comuna"].duplicated().sum()

print("Duplicados encontrados:", duplicados)

if duplicados == 0:
    print("RESULTADO: CORRECTO")
else:
    print("RESULTADO: REVISAR")


# 4. Verificar valores faltantes
print("\n3. VALORES FALTANTES")

columnas_principales = [
    "codigo_comuna",
    "comuna_oficial",
    "poblacion_2024",
    "informes_no_anulados",
    "frecuencia_100k_hab_anio"
]

faltantes = datos[columnas_principales].isna().sum()

print(faltantes.to_string())

if faltantes.sum() == 0:
    print("RESULTADO: CORRECTO")
else:
    print("RESULTADO: REVISAR")


# 5. Verificar la frecuencia documental normalizada
print("\n4. VALIDACIÓN DE FRECUENCIA NORMALIZADA")

meses_observados = 33

# Fórmula: informes / población × 100.000 × (12 / 33)
tasa_recalculada = (
    datos["informes_no_anulados"]
    / datos["poblacion_2024"]
    * 100000
    * (12 / meses_observados)
)

# Comparar con la tasa almacenada en F3
comparacion = np.isclose(
    tasa_recalculada,
    datos["frecuencia_100k_hab_anio"],
    atol=0.011,
    rtol=0
)

correctas = int(comparacion.sum())

print("Tasas verificadas:", correctas, "de", len(datos))

if correctas == len(datos):
    print("RESULTADO: CORRECTO")
else:
    print("RESULTADO: REVISAR")

    print("\nComunas con diferencias:")

    diferencias = datos.loc[
        ~comparacion,
        [
            "comuna_oficial",
            "informes_no_anulados",
            "poblacion_2024",
            "frecuencia_100k_hab_anio"
        ]
    ].copy()

    diferencias["tasa_recalculada"] = (
        tasa_recalculada.loc[~comparacion]
    )

    display(diferencias)


# 6. Resumen de la validación
print("\n" + "=" * 65)
print("RESUMEN DE VALIDACIÓN")
print("=" * 65)

validaciones = {
    "Cantidad de 15 comunas": len(datos) == 15,
    "Sin comunas duplicadas": duplicados == 0,
    "Sin faltantes principales": faltantes.sum() == 0,
    "Tasas normalizadas consistentes": correctas == len(datos)
}

for nombre, resultado in validaciones.items():
    estado = "CORRECTO" if resultado else "REVISAR"
    print(f"{nombre}: {estado}")

print("\n" + "=" * 65)

if all(validaciones.values()):
    print("VALIDACIÓN INICIAL COMPLETADA CORRECTAMENTE")
else:
    print("VALIDACIÓN INICIAL CON OBSERVACIONES")

print("=" * 65)
```

### 3.7. Verificación de indicadores de recurrencia causal, duración documental y cobertura temporal

La recurrencia documental no se interpretará como evidencia suficiente para establecer relaciones causales físicas entre las fallas y las comunas.

#### 3.7.3. Verificación de duración documental

El análisis temporal considera los intervalos documentales que cumplen los criterios de elegibilidad establecidos durante F2 y F3.

Las variables examinadas son:

- `informes_validos`: cantidad de registros elegibles informada en el panel.
- `mediana_horas`: mediana de los intervalos documentales elegibles.
- `percentil_75_horas`: percentil 75 de dichos intervalos.

La validación comprobará que los valores temporales sean no negativos y que la mediana no supere el percentil 75.

Esta comprobación permite identificar inconsistencias básicas entre las estadísticas descriptivas, pero no sustituye la verificación de los intervalos originales.

#### 3.7.4. Verificación de cobertura temporal

La cobertura temporal constituye un indicador de disponibilidad de información elegible para el análisis de intervalos documentales.

Se comprobará si el porcentaje informado en `cobertura_temporal_pct` es consistente con la relación entre `informes_validos` y `asociaciones_documentales`.

La verificación permitirá identificar el denominador utilizado y comprobar que los porcentajes se encuentren dentro del intervalo de 0 % a 100 %.

Se mantendrá la distinción entre informes documentales únicos y asociaciones informe–comuna, evitando atribuir a los indicadores un significado distinto del definido durante las fases anteriores.

#### 3.7.5. Procedimiento de validación

Se utilizarán Pandas y NumPy para realizar las comprobaciones sobre una copia del panel integrado.

La comparación de porcentajes considerará una tolerancia absoluta de 0,011 puntos porcentuales, debido a las posibles diferencias de redondeo.

Los resultados se presentarán en una tabla de verificación por comuna y en un resumen de cumplimiento.

La validación realizada comprobará la consistencia interna de los indicadores. La correspondencia íntegra con los registros procesados de F2 y los procedimientos de F3 requerirá verificaciones complementarias.

```python
# ============================================================
# MCDI500 - GRUPO 7
# FASE 4 - APARTADO 3.7
# VALIDACIÓN DE RECURRENCIA, DURACIÓN Y COBERTURA TEMPORAL
# ============================================================

import pandas as pd
import numpy as np
from IPython.display import display

print("=" * 65)
print("F4 - VALIDACIÓN DE INDICADORES COMPLEMENTARIOS")
print("=" * 65)

# ------------------------------------------------------------
# 1. TRABAJAR CON UNA COPIA DEL PANEL DE F3
# ------------------------------------------------------------

datos_validacion = panel_f3.copy()

# Verificar que existan las columnas necesarias
columnas_requeridas = [
    "codigo_comuna",
    "comuna_oficial",
    "asociaciones_documentales",
    "causa_mas_recurrente",
    "informes_causa_principal",
    "porcentaje_causa_principal",
    "informes_validos",
    "mediana_horas",
    "percentil_75_horas",
    "cobertura_temporal_pct"
]

columnas_faltantes = [
    columna
    for columna in columnas_requeridas
    if columna not in datos_validacion.columns
]

if columnas_faltantes:
    raise ValueError(
        "Faltan columnas en el panel: "
        + ", ".join(columnas_faltantes)
    )

# ------------------------------------------------------------
# 2. PREPARAR LAS VARIABLES NUMÉRICAS
# ------------------------------------------------------------

columnas_numericas = [
    "asociaciones_documentales",
    "informes_causa_principal",
    "porcentaje_causa_principal",
    "informes_validos",
    "mediana_horas",
    "percentil_75_horas",
    "cobertura_temporal_pct"
]

for columna in columnas_numericas:
    datos_validacion[columna] = pd.to_numeric(
        datos_validacion[columna],
        errors="coerce"
    )

# ------------------------------------------------------------
# 3. VALIDAR LA RECURRENCIA CAUSAL
# ------------------------------------------------------------

print("\n1. VALIDACIÓN DE RECURRENCIA CAUSAL")

asociaciones = datos_validacion["asociaciones_documentales"]

conteos_causales = datos_validacion["informes_causa_principal"]

porcentajes_causales = datos_validacion[
    "porcentaje_causa_principal"
]

# Verificar que el denominador sea positivo
denominador_valido = asociaciones.gt(0)

# Calcular el porcentaje a partir de las asociaciones
porcentaje_causal_calculado = (
    conteos_causales
    .div(asociaciones.where(denominador_valido))
    .mul(100)
)

# Comparar porcentajes con tolerancia de redondeo
causa_porcentaje_coherente = np.isclose(
    porcentaje_causal_calculado,
    porcentajes_causales,
    atol=0.011,
    rtol=0,
    equal_nan=False
)

# Verificar que el conteo no supere el denominador
causa_conteo_valido = (
    conteos_causales.ge(0)
    & conteos_causales.le(asociaciones)
)

# Verificar presencia de descripción causal
causa_descripcion_valida = (
    datos_validacion["causa_mas_recurrente"]
    .fillna("")
    .astype(str)
    .str.strip()
    .ne("")
)

validacion_causal = (
    denominador_valido
    & causa_porcentaje_coherente
    & causa_conteo_valido
    & causa_descripcion_valida
)

print(
    "Comunas con recurrencia consistente:",
    int(validacion_causal.sum()),
    "de",
    len(datos_validacion)
)

# ------------------------------------------------------------
# 4. VALIDAR LA DURACIÓN DOCUMENTAL
# ------------------------------------------------------------

print("\n2. VALIDACIÓN DE DURACIÓN DOCUMENTAL")

mediana = datos_validacion["mediana_horas"]

percentil_75 = datos_validacion["percentil_75_horas"]

registros_validos = datos_validacion["informes_validos"]

# Cuando existen registros elegibles, se requieren estadísticas
duracion_con_registros = (
    registros_validos.gt(0)
    & mediana.notna()
    & percentil_75.notna()
    & mediana.ge(0)
    & percentil_75.ge(mediana)
)

# Cuando no existen registros, se admiten ambas estadísticas nulas
duracion_sin_registros = (
    registros_validos.eq(0)
    & mediana.isna()
    & percentil_75.isna()
)

validacion_duracion = (
    duracion_con_registros
    | duracion_sin_registros
)

print(
    "Comunas con estadísticas temporales consistentes:",
    int(validacion_duracion.sum()),
    "de",
    len(datos_validacion)
)

# ------------------------------------------------------------
# 5. VALIDAR LA COBERTURA TEMPORAL
# ------------------------------------------------------------

print("\n3. VALIDACIÓN DE COBERTURA TEMPORAL")

cobertura_reportada = datos_validacion[
    "cobertura_temporal_pct"
]

# Comprobar la relación con asociaciones documentales
cobertura_calculada = (
    registros_validos
    .div(asociaciones.where(denominador_valido))
    .mul(100)
)

cobertura_porcentaje_coherente = np.isclose(
    cobertura_calculada,
    cobertura_reportada,
    atol=0.011,
    rtol=0,
    equal_nan=False
)

cobertura_rango_valido = (
    cobertura_reportada.ge(0)
    & cobertura_reportada.le(100)
    & registros_validos.ge(0)
    & registros_validos.le(asociaciones)
)

validacion_cobertura = (
    denominador_valido
    & cobertura_porcentaje_coherente
    & cobertura_rango_valido
)

print(
    "Comunas con cobertura temporal consistente:",
    int(validacion_cobertura.sum()),
    "de",
    len(datos_validacion)
)

# ------------------------------------------------------------
# 6. CONSTRUIR TABLA DE VALIDACIÓN
# ------------------------------------------------------------

tabla_validacion = pd.DataFrame({
    "Comuna": datos_validacion["comuna_oficial"],
    "Recurrencia": np.where(
        validacion_causal,
        "CORRECTO",
        "REVISAR"
    ),
    "Duración": np.where(
        validacion_duracion,
        "CORRECTO",
        "REVISAR"
    ),
    "Cobertura": np.where(
        validacion_cobertura,
        "CORRECTO",
        "REVISAR"
    )
})

tabla_validacion["Estado_general"] = np.where(
    (
        validacion_causal
        & validacion_duracion
        & validacion_cobertura
    ),
    "CORRECTO",
    "REVISAR"
)

# Mostrar únicamente las comunas que requieren revisión
observaciones = tabla_validacion.loc[
    tabla_validacion["Estado_general"].eq("REVISAR")
]

# ------------------------------------------------------------
# 7. RESUMEN FINAL
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("RESUMEN DE VALIDACIÓN")
print("=" * 65)

resultados = {
    "Recurrencia causal": bool(validacion_causal.all()),
    "Duración documental": bool(validacion_duracion.all()),
    "Cobertura temporal": bool(validacion_cobertura.all())
}

for nombre, resultado in resultados.items():
    estado = "CORRECTO" if resultado else "REVISAR"
    print(f"{nombre}: {estado}")

print(
    "\nComunas con las tres verificaciones correctas:",
    int(tabla_validacion["Estado_general"].eq("CORRECTO").sum()),
    "de",
    len(tabla_validacion)
)

if observaciones.empty:
    print("\nNo se identificaron inconsistencias internas.")
else:
    print("\nCOMUNAS QUE REQUIEREN REVISIÓN:")
    display(observaciones)

print("\n" + "=" * 65)

if all(resultados.values()):
    print("VALIDACIÓN COMPLEMENTARIA COMPLETADA CORRECTAMENTE")
else:
    print("VALIDACIÓN COMPLEMENTARIA CON OBSERVACIONES")

print("=" * 65)
```

### 5.1. Visualización de la frecuencia documental normalizada por comuna

#### 5.1.1. Propósito de la visualización

La primera visualización tiene como propósito representar y comparar la frecuencia documental normalizada de los informes no anulados del Coordinador Eléctrico Nacional (CEN) asociados con las 15 comunas más pobladas de Chile.

La comparación utiliza los indicadores obtenidos durante F3 y consolidados en el apartado anterior, considerando la población de referencia del Censo 2024 y el período documental de 33 meses.

El objetivo consiste en facilitar la identificación de diferencias territoriales mediante una representación gráfica que permita comparar las comunas utilizando una misma unidad de medida.

#### 5.1.2. Variables utilizadas

La visualización se construirá utilizando las siguientes variables del panel integrado:

| Variable | Descripción | Utilización |
|---|---|---|
| `comuna_oficial` | Nombre oficial de la comuna. | Identificación de las categorías del gráfico. |
| `frecuencia_100k_hab_anio` | Frecuencia documental normalizada por 100.000 habitantes-año. | Magnitud representada en el eje horizontal. |

Los indicadores utilizados corresponden a los valores calculados y verificados previamente. No se realizarán nuevos cálculos de frecuencia durante la construcción del gráfico.

#### 5.1.3. Selección del tipo de visualización

Se utilizará un **gráfico de barras horizontales**, construido mediante la biblioteca Matplotlib.

Esta representación se considera pertinente debido a que permite comparar una variable cuantitativa entre múltiples categorías territoriales, manteniendo la legibilidad de los nombres de las comunas.

Las comunas se presentarán ordenadas de mayor a menor frecuencia documental normalizada, incorporando etiquetas numéricas que permitan identificar sus valores.

El eje horizontal representará la frecuencia documental por 100.000 habitantes-año, mientras que el eje vertical identificará las 15 comunas.

#### 5.1.4. Consideraciones metodológicas

La frecuencia documental normalizada representa una medida descriptiva construida a partir de informes no anulados, población de referencia y período de observación.

Por lo tanto, las diferencias representadas en el gráfico no deben interpretarse automáticamente como diferencias en la calidad del suministro eléctrico, probabilidad individual de interrupción o indicadores regulatorios de continuidad del servicio.

La visualización conservará los valores provenientes del núcleo algorítmico de F3 y permitirá su comparación bajo una escala común.

#### 5.1.5. Resultado esperado

Se espera obtener una representación de las 15 comunas ordenadas según su frecuencia documental normalizada, que permita identificar los mayores y menores valores y establecer una base visual para la interpretación de los resultados.

La figura será exportada en formato PNG, con resolución adecuada para su utilización en el informe técnico final y la presentación audiovisual del proyecto.

```python
# ============================================================
# MCDI500 - GRUPO 7
# FASE 4 - APARTADO 5.1
# VISUALIZACIÓN DE FRECUENCIA DOCUMENTAL NORMALIZADA
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

print("=" * 65)
print("F4 - GRÁFICO 5.1: FRECUENCIA DOCUMENTAL NORMALIZADA")
print("=" * 65)

# ------------------------------------------------------------
# 1. RECUPERAR LOS INDICADORES CONSOLIDADOS
# ------------------------------------------------------------

# Utilizamos la tabla generada en el apartado 4.
# No se modifican ni recalculan los indicadores de F3.

datos_grafico = tabla_frecuencia[
    [
        "comuna_oficial",
        "frecuencia_100k_hab_anio"
    ]
].copy()

# ------------------------------------------------------------
# 2. VERIFICAR LOS DATOS DE ENTRADA
# ------------------------------------------------------------

if len(datos_grafico) != 15:
    raise ValueError(
        "El gráfico requiere exactamente 15 comunas."
    )

if datos_grafico["comuna_oficial"].duplicated().any():
    raise ValueError(
        "Se identificaron comunas duplicadas."
    )

if datos_grafico.isna().any().any():
    raise ValueError(
        "Existen valores faltantes en los datos del gráfico."
    )

if not pd.api.types.is_numeric_dtype(
    datos_grafico["frecuencia_100k_hab_anio"]
):
    raise TypeError(
        "La frecuencia documental debe ser numérica."
    )

if not np.isfinite(
    datos_grafico["frecuencia_100k_hab_anio"]
).all():
    raise ValueError(
        "Existen valores de frecuencia no finitos."
    )

if (
    datos_grafico["frecuencia_100k_hab_anio"] < 0
).any():
    raise ValueError(
        "Se identificaron frecuencias negativas."
    )

print("\nValidación de entrada: CORRECTA")
print("Comunas:", len(datos_grafico))

# ------------------------------------------------------------
# 3. ORDENAR LAS COMUNAS PARA LA REPRESENTACIÓN
# ------------------------------------------------------------

# En un gráfico de barras horizontales, el orden
# ascendente permite visualizar el mayor valor arriba.

datos_grafico = datos_grafico.sort_values(
    by="frecuencia_100k_hab_anio",
    ascending=True,
    kind="mergesort"
).reset_index(drop=True)

comunas = datos_grafico["comuna_oficial"]

frecuencias = datos_grafico[
    "frecuencia_100k_hab_anio"
]

# ------------------------------------------------------------
# 4. CONSTRUIR LA VISUALIZACIÓN
# ------------------------------------------------------------

fig, ax = plt.subplots(
    figsize=(12, 8)
)

barras = ax.barh(
    comunas,
    frecuencias,
    height=0.72,
    color="#2864A5",
    edgecolor="white",
    linewidth=0.6
)

# ------------------------------------------------------------
# 5. CONFIGURAR TÍTULOS Y EJES
# ------------------------------------------------------------

ax.set_title(
    "Frecuencia documental normalizada por comuna\n"
    "15 comunas más pobladas de Chile | 2024–2026",
    fontsize=15,
    fontweight="bold",
    pad=18
)

ax.set_xlabel(
    "Informes no anulados por 100.000 habitantes-año",
    fontsize=11,
    labelpad=12
)

ax.set_ylabel(
    "Comuna",
    fontsize=11,
    labelpad=12
)

ax.tick_params(
    axis="both",
    labelsize=10
)

# La escala comienza en cero para no exagerar diferencias.
ax.set_xlim(
    0,
    float(frecuencias.max()) * 1.18
)

# Cuadrícula de referencia.
ax.grid(
    axis="x",
    linestyle="--",
    alpha=0.25
)

ax.set_axisbelow(True)

# Eliminar bordes visualmente innecesarios.
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

# ------------------------------------------------------------
# 6. INCORPORAR ETIQUETAS NUMÉRICAS
# ------------------------------------------------------------

desplazamiento = float(
    frecuencias.max()
) * 0.012

for barra, valor in zip(
    barras,
    frecuencias
):

    etiqueta = f"{valor:.2f}".replace(".", ",")

    ax.text(
        barra.get_width() + desplazamiento,
        barra.get_y() + barra.get_height() / 2,
        etiqueta,
        va="center",
        ha="left",
        fontsize=9,
        fontweight="medium"
    )

# ------------------------------------------------------------
# 7. INCORPORAR FUENTE Y NOTA METODOLÓGICA
# ------------------------------------------------------------

fig.text(
    0.12,
    0.035,
    "Fuente: elaboración propia a partir de registros CEN "
    "y población INE, Censo 2024.",
    fontsize=9,
    ha="left"
)

fig.text(
    0.12,
    0.015,
    "Nota: frecuencia documental normalizada; "
    "no corresponde a SAIFI ni a probabilidad de interrupción.",
    fontsize=8.5,
    ha="left",
    color="#555555"
)

plt.tight_layout(
    rect=[0, 0.075, 1, 1]
)

# ------------------------------------------------------------
# 8. EXPORTAR LA FIGURA
# ------------------------------------------------------------

directorio_figuras_f4 = (
    raiz_proyecto / "F4" / "figuras"
)

directorio_figuras_f4.mkdir(
    parents=True,
    exist_ok=True
)

ruta_figura = (
    directorio_figuras_f4 /
    "F4_Figura_501_Frecuencia_Comunal.png"
)

fig.savefig(
    ruta_figura,
    dpi=300,
    bbox_inches="tight",
    facecolor="white"
)

# ------------------------------------------------------------
# 9. MOSTRAR EL GRÁFICO
# ------------------------------------------------------------

plt.show()

# ------------------------------------------------------------
# 10. REGISTRAR RESULTADOS PRINCIPALES
# ------------------------------------------------------------

comuna_mayor = datos_grafico.iloc[-1]
comuna_menor = datos_grafico.iloc[0]

print("\n" + "=" * 65)
print("RESUMEN DE LA VISUALIZACIÓN")
print("=" * 65)

print(
    "Comunas representadas:",
    len(datos_grafico)
)

print("\nMayor frecuencia documental:")
print(
    f"{comuna_mayor['comuna_oficial']}: "
    f"{comuna_mayor['frecuencia_100k_hab_anio']:.2f}"
)

print("\nMenor frecuencia documental:")
print(
    f"{comuna_menor['comuna_oficial']}: "
    f"{comuna_menor['frecuencia_100k_hab_anio']:.2f}"
)

print("\nFigura exportada:")
print(ruta_figura)

print("\n" + "=" * 65)
print("GRÁFICO 5.1 GENERADO CORRECTAMENTE")
print("=" * 65)
```

```python
# ============================================================
# MCDI500 - GRUPO 7
# FASE 4 - APARTADO 5.2
# VISUALIZACIÓN DE RECURRENCIA CAUSAL DOCUMENTAL
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from pathlib import Path
from matplotlib.patches import Patch
from textwrap import fill


print("=" * 65)
print("F4 - GRÁFICO 5.2: RECURRENCIA CAUSAL DOCUMENTAL")
print("=" * 65)


# ------------------------------------------------------------
# 1. RECUPERAR LOS INDICADORES CONSOLIDADOS
# ------------------------------------------------------------

# Utilizamos la Tabla 4.2 generada anteriormente.
# No modificamos los datos ni recalculamos los indicadores.

datos_grafico = tabla_recurrencia[
    [
        "comuna_oficial",
        "asociaciones_documentales",
        "causa_mas_recurrente",
        "informes_causa_principal",
        "porcentaje_causa_principal"
    ]
].copy()


# ------------------------------------------------------------
# 2. VERIFICAR LOS DATOS DE ENTRADA
# ------------------------------------------------------------

if len(datos_grafico) != 15:
    raise ValueError(
        "La visualización requiere exactamente 15 comunas."
    )

if datos_grafico["comuna_oficial"].duplicated().any():
    raise ValueError(
        "Se identificaron comunas duplicadas."
    )

if datos_grafico.isna().any().any():
    raise ValueError(
        "Existen valores faltantes en los datos."
    )

columnas_numericas = [
    "asociaciones_documentales",
    "informes_causa_principal",
    "porcentaje_causa_principal"
]

for columna in columnas_numericas:

    if not pd.api.types.is_numeric_dtype(
        datos_grafico[columna]
    ):
        raise TypeError(
            f"La columna {columna} debe ser numérica."
        )

    if not np.isfinite(
        datos_grafico[columna]
    ).all():
        raise ValueError(
            f"La columna {columna} contiene valores no finitos."
        )

if (
    datos_grafico["porcentaje_causa_principal"] < 0
).any() or (
    datos_grafico["porcentaje_causa_principal"] > 100
).any():

    raise ValueError(
        "Los porcentajes deben encontrarse entre 0 y 100."
    )

if (
    datos_grafico["asociaciones_documentales"] <= 0
).any():

    raise ValueError(
        "Las asociaciones documentales deben ser positivas."
    )

if (
    datos_grafico["informes_causa_principal"] < 0
).any() or (
    datos_grafico["informes_causa_principal"]
    > datos_grafico["asociaciones_documentales"]
).any():

    raise ValueError(
        "Los conteos causales no son consistentes."
    )

print("\nValidación de entrada: CORRECTA")
print("Comunas identificadas:", len(datos_grafico))


# ------------------------------------------------------------
# 3. ORDENAR LOS RESULTADOS
# ------------------------------------------------------------

# Se utiliza orden ascendente para que el mayor porcentaje
# aparezca en la parte superior del gráfico horizontal.

datos_grafico = datos_grafico.sort_values(
    by="porcentaje_causa_principal",
    ascending=True,
    kind="mergesort"
).reset_index(drop=True)

comunas = datos_grafico["comuna_oficial"]

porcentajes = datos_grafico[
    "porcentaje_causa_principal"
]

categorias = datos_grafico[
    "causa_mas_recurrente"
]


# ------------------------------------------------------------
# 4. IDENTIFICAR LAS CATEGORÍAS DOCUMENTALES
# ------------------------------------------------------------

# Conservamos las denominaciones originales provenientes
# de F3, sin modificar ni reclasificar las categorías.

categorias_unicas = sorted(
    categorias.unique()
)

paleta = plt.get_cmap("tab10")

colores_categorias = {
    categoria: paleta(indice % 10)
    for indice, categoria in enumerate(categorias_unicas)
}

colores_barras = [
    colores_categorias[categoria]
    for categoria in categorias
]

print("\nDescripciones causales predominantes:")

for categoria in categorias_unicas:
    print("-", categoria)


# ------------------------------------------------------------
# 5. CONSTRUIR EL GRÁFICO
# ------------------------------------------------------------

fig, ax = plt.subplots(
    figsize=(16, 9)
)

barras = ax.barh(
    comunas,
    porcentajes,
    color=colores_barras,
    height=0.72,
    edgecolor="white",
    linewidth=0.6
)


# ------------------------------------------------------------
# 6. CONFIGURAR TÍTULO, EJES Y ESCALA
# ------------------------------------------------------------

ax.set_title(
    "Recurrencia de la descripción causal predominante\n"
    "15 comunas más pobladas de Chile | 2024–2026",
    fontsize=15,
    fontweight="bold",
    pad=18
)

ax.set_xlabel(
    "Participación de la descripción predominante (%)",
    fontsize=11,
    labelpad=12
)

ax.set_ylabel(
    "Comuna",
    fontsize=11,
    labelpad=12
)

ax.tick_params(
    axis="both",
    labelsize=10
)

# El eje porcentual comienza en cero.
# Se incorpora espacio para las etiquetas.

limite_x = max(
    35,
    float(porcentajes.max()) * 1.35
)

ax.set_xlim(
    0,
    limite_x
)

ax.grid(
    axis="x",
    linestyle="--",
    alpha=0.25
)

ax.set_axisbelow(True)

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)


# ------------------------------------------------------------
# 7. INCORPORAR ETIQUETAS NUMÉRICAS
# ------------------------------------------------------------

# Cada barra incorpora:
# porcentaje (conteo causal / asociaciones documentales).

desplazamiento = (
    float(porcentajes.max()) * 0.015
)

for barra, (_, fila) in zip(
    barras,
    datos_grafico.iterrows()
):

    porcentaje = fila[
        "porcentaje_causa_principal"
    ]

    conteo = int(
        fila["informes_causa_principal"]
    )

    asociaciones = int(
        fila["asociaciones_documentales"]
    )

    etiqueta = (
        f"{porcentaje:.2f}% "
        f"({conteo}/{asociaciones})"
    ).replace(".", ",")

    ax.text(
        barra.get_width() + desplazamiento,
        barra.get_y() + barra.get_height() / 2,
        etiqueta,
        va="center",
        ha="left",
        fontsize=9
    )


# ------------------------------------------------------------
# 8. CONSTRUIR LA LEYENDA DE CATEGORÍAS
# ------------------------------------------------------------

# Las descripciones largas se distribuyen en varias líneas
# para mejorar su lectura.

elementos_leyenda = []

for categoria in categorias_unicas:

    etiqueta_leyenda = fill(
        categoria,
        width=43
    )

    elementos_leyenda.append(
        Patch(
            facecolor=colores_categorias[categoria],
            label=etiqueta_leyenda
        )
    )

ax.legend(
    handles=elementos_leyenda,
    title="Descripción causal predominante",
    title_fontsize=10,
    fontsize=9,
    loc="upper left",
    bbox_to_anchor=(1.02, 1),
    frameon=False
)


# ------------------------------------------------------------
# 9. INCORPORAR FUENTE Y NOTA METODOLÓGICA
# ------------------------------------------------------------

fig.text(
    0.10,
    0.045,
    "Fuente: elaboración propia a partir de registros "
    "documentales CEN procesados en F2 y F3.",
    fontsize=9,
    ha="left"
)

fig.text(
    0.10,
    0.025,
    "Nota: se representa únicamente la descripción "
    "predominante de cada comuna. "
    "Los porcentajes utilizan asociaciones documentales.",
    fontsize=8.5,
    ha="left",
    color="#555555"
)

# Reservar espacio a la derecha para la leyenda.
plt.subplots_adjust(
    left=0.14,
    right=0.66,
    top=0.87,
    bottom=0.15
)


# ------------------------------------------------------------
# 10. EXPORTAR EL GRÁFICO
# ------------------------------------------------------------

directorio_figuras_f4 = (
    raiz_proyecto / "F4" / "figuras"
)

directorio_figuras_f4.mkdir(
    parents=True,
    exist_ok=True
)

ruta_figura = (
    directorio_figuras_f4 /
    "F4_Figura_502_Recurrencia_Causal.png"
)

fig.savefig(
    ruta_figura,
    dpi=300,
    bbox_inches="tight",
    facecolor="white"
)


# ------------------------------------------------------------
# 11. MOSTRAR LA VISUALIZACIÓN
# ------------------------------------------------------------

plt.show()


# ------------------------------------------------------------
# 12. RESUMEN DE RESULTADOS
# ------------------------------------------------------------

comuna_mayor = datos_grafico.iloc[-1]

comuna_menor = datos_grafico.iloc[0]

print("\n" + "=" * 65)
print("RESUMEN DE LA VISUALIZACIÓN")
print("=" * 65)

print(
    "Comunas representadas:",
    len(datos_grafico)
)

print(
    "Descripciones predominantes distintas:",
    len(categorias_unicas)
)

print("\nMayor participación porcentual:")

print(
    f"{comuna_mayor['comuna_oficial']}: "
    f"{comuna_mayor['porcentaje_causa_principal']:.2f}%"
)

print(
    "Descripción:",
    comuna_mayor["causa_mas_recurrente"]
)

print("\nMenor participación porcentual:")

print(
    f"{comuna_menor['comuna_oficial']}: "
    f"{comuna_menor['porcentaje_causa_principal']:.2f}%"
)

print(
    "Descripción:",
    comuna_menor["causa_mas_recurrente"]
)

print("\nFigura exportada:")
print(ruta_figura)

print("\n" + "=" * 65)
print("GRÁFICO 5.2 GENERADO CORRECTAMENTE")
print("=" * 65)
```

```python
# ============================================================
# MCDI500 - GRUPO 7
# FASE 4 - APARTADO 5.3
# VISUALIZACIÓN DE DURACIÓN DOCUMENTAL Y COBERTURA TEMPORAL
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from pathlib import Path
from matplotlib.transforms import blended_transform_factory


print("=" * 70)
print("F4 - GRÁFICO 5.3: DURACIÓN Y COBERTURA TEMPORAL")
print("=" * 70)


# ------------------------------------------------------------
# 1. RECUPERAR LOS INDICADORES CONSOLIDADOS
# ------------------------------------------------------------

# Se utiliza la Tabla 4.3 generada anteriormente.
# Los indicadores de F3 no se modifican ni recalculan.

datos_grafico = tabla_duracion[
    [
        "comuna_oficial",
        "informes_validos",
        "mediana_horas",
        "percentil_75_horas",
        "cobertura_temporal_pct"
    ]
].copy()


# ------------------------------------------------------------
# 2. VALIDAR LOS DATOS DE ENTRADA
# ------------------------------------------------------------

if len(datos_grafico) != 15:
    raise ValueError(
        "La visualización requiere exactamente 15 comunas."
    )

if datos_grafico["comuna_oficial"].duplicated().any():
    raise ValueError(
        "Se identificaron comunas duplicadas."
    )

if datos_grafico.isna().any().any():
    raise ValueError(
        "Existen valores faltantes en los datos."
    )

columnas_numericas = [
    "informes_validos",
    "mediana_horas",
    "percentil_75_horas",
    "cobertura_temporal_pct"
]

for columna in columnas_numericas:

    datos_grafico[columna] = pd.to_numeric(
        datos_grafico[columna],
        errors="raise"
    )

    if not np.isfinite(datos_grafico[columna]).all():
        raise ValueError(
            f"Se identificaron valores no finitos en {columna}."
        )

if (datos_grafico["informes_validos"] <= 0).any():
    raise ValueError(
        "Existen comunas sin registros temporales elegibles."
    )

if (datos_grafico["mediana_horas"] < 0).any():
    raise ValueError(
        "Se identificaron medianas negativas."
    )

if (
    datos_grafico["percentil_75_horas"]
    < datos_grafico["mediana_horas"]
).any():
    raise ValueError(
        "El percentil 75 no puede ser inferior a la mediana."
    )

if (
    (datos_grafico["cobertura_temporal_pct"] < 0)
    |
    (datos_grafico["cobertura_temporal_pct"] > 100)
).any():
    raise ValueError(
        "La cobertura temporal debe estar entre 0 y 100 %."
    )

print("\nValidación de entrada: CORRECTA")
print("Comunas identificadas:", len(datos_grafico))


# ------------------------------------------------------------
# 3. ORDENAR LOS RESULTADOS
# ------------------------------------------------------------

# Se ordenan las comunas de mayor a menor mediana.
# El mayor valor aparecerá en la parte superior.

datos_grafico = datos_grafico.sort_values(
    by="mediana_horas",
    ascending=False,
    kind="mergesort"
).reset_index(drop=True)

comunas = datos_grafico["comuna_oficial"]

medianas = datos_grafico["mediana_horas"]

percentiles_75 = datos_grafico["percentil_75_horas"]

coberturas = datos_grafico["cobertura_temporal_pct"]

registros_elegibles = datos_grafico["informes_validos"]


# ------------------------------------------------------------
# 4. PREPARAR LA POSICIÓN DE LAS BARRAS
# ------------------------------------------------------------

posiciones = np.arange(len(datos_grafico))

ancho_barras = 0.34

posiciones_mediana = posiciones - 0.19

posiciones_percentil = posiciones + 0.19


# ------------------------------------------------------------
# 5. CONSTRUIR EL GRÁFICO
# ------------------------------------------------------------

fig, ax = plt.subplots(
    figsize=(16, 9)
)

barras_mediana = ax.barh(
    posiciones_mediana,
    medianas,
    height=ancho_barras,
    color="#2864A5",
    edgecolor="white",
    linewidth=0.5,
    label="Mediana documental"
)

barras_percentil = ax.barh(
    posiciones_percentil,
    percentiles_75,
    height=ancho_barras,
    color="#E69F00",
    edgecolor="white",
    linewidth=0.5,
    label="Percentil 75"
)


# ------------------------------------------------------------
# 6. CONFIGURAR LAS COMUNAS Y SU ORDEN
# ------------------------------------------------------------

ax.set_yticks(
    posiciones
)

ax.set_yticklabels(
    comunas,
    fontsize=10
)

# Colocar la comuna con mayor mediana arriba.
ax.invert_yaxis()


# ------------------------------------------------------------
# 7. CONFIGURAR TÍTULO, EJES Y ESCALA
# ------------------------------------------------------------

ax.set_title(
    "Intervalos documentales elegibles por comuna\n"
    "Mediana, percentil 75 y cobertura temporal | 2024–2026",
    fontsize=15,
    fontweight="bold",
    pad=20
)

ax.set_xlabel(
    "Intervalo documental (horas)",
    fontsize=11,
    labelpad=12
)

ax.set_ylabel(
    "Comuna",
    fontsize=11,
    labelpad=12
)

ax.tick_params(
    axis="x",
    labelsize=10
)

# El eje comienza en cero y deja espacio para las etiquetas.
valor_maximo = float(
    percentiles_75.max()
)

limite_x = max(
    1,
    valor_maximo * 1.22
)

ax.set_xlim(
    0,
    limite_x
)

ax.grid(
    axis="x",
    linestyle="--",
    alpha=0.25
)

ax.set_axisbelow(True)

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)


# ------------------------------------------------------------
# 8. INCORPORAR ETIQUETAS DE DURACIÓN
# ------------------------------------------------------------

desplazamiento = max(
    0.04,
    valor_maximo * 0.015
)

for barra, valor in zip(
    barras_mediana,
    medianas
):

    etiqueta = f"{valor:.2f}".replace(".", ",")

    ax.text(
        barra.get_width() + desplazamiento,
        barra.get_y() + barra.get_height() / 2,
        etiqueta,
        va="center",
        ha="left",
        fontsize=8.5
    )


for barra, valor in zip(
    barras_percentil,
    percentiles_75
):

    etiqueta = f"{valor:.2f}".replace(".", ",")

    ax.text(
        barra.get_width() + desplazamiento,
        barra.get_y() + barra.get_height() / 2,
        etiqueta,
        va="center",
        ha="left",
        fontsize=8.5
    )


# ------------------------------------------------------------
# 9. INCORPORAR COBERTURA Y REGISTROS ELEGIBLES
# ------------------------------------------------------------

# La cobertura se representa como una columna textual,
# separada de la escala de duración en horas.
#
# De esta manera evitamos utilizar un mismo eje
# para magnitudes diferentes.

transformacion = blended_transform_factory(
    ax.transAxes,
    ax.transData
)

ax.text(
    1.04,
    1.025,
    "Cobertura temporal\n(% y registros elegibles)",
    transform=ax.transAxes,
    fontsize=10,
    fontweight="bold",
    ha="left",
    va="bottom"
)

for posicion, cobertura, cantidad in zip(
    posiciones,
    coberturas,
    registros_elegibles
):

    texto_cobertura = (
        f"{cobertura:.2f}%  |  n = {int(cantidad)}"
    ).replace(".", ",")

    ax.text(
        1.04,
        posicion,
        texto_cobertura,
        transform=transformacion,
        ha="left",
        va="center",
        fontsize=9
    )


# ------------------------------------------------------------
# 10. CONFIGURAR LEYENDA
# ------------------------------------------------------------

ax.legend(
    loc="upper center",
    bbox_to_anchor=(0.5, -0.075),
    ncol=2,
    fontsize=10,
    frameon=False
)


# ------------------------------------------------------------
# 11. INCORPORAR FUENTE Y NOTA METODOLÓGICA
# ------------------------------------------------------------

fig.text(
    0.10,
    0.065,
    "Fuente: elaboración propia a partir de registros "
    "documentales CEN procesados en F2 y F3.",
    fontsize=9,
    ha="left"
)

fig.text(
    0.10,
    0.040,
    "Nota: intervalos entre hitos documentales elegibles; "
    "no equivalen necesariamente a duración efectiva de cortes eléctricos.",
    fontsize=8.5,
    ha="left",
    color="#555555"
)

fig.text(
    0.10,
    0.018,
    "La cobertura corresponde al porcentaje de asociaciones "
    "documentales con intervalos elegibles.",
    fontsize=8.5,
    ha="left",
    color="#555555"
)


# ------------------------------------------------------------
# 12. AJUSTAR LA DISTRIBUCIÓN DE LA FIGURA
# ------------------------------------------------------------

# Se reserva espacio a la derecha para mostrar
# cobertura temporal y cantidad de registros.

plt.subplots_adjust(
    left=0.14,
    right=0.74,
    top=0.84,
    bottom=0.20
)


# ------------------------------------------------------------
# 13. EXPORTAR LA FIGURA
# ------------------------------------------------------------

directorio_figuras_f4 = (
    raiz_proyecto / "F4" / "figuras"
)

directorio_figuras_f4.mkdir(
    parents=True,
    exist_ok=True
)

ruta_figura = (
    directorio_figuras_f4 /
    "F4_Figura_503_Duracion_Cobertura.png"
)

fig.savefig(
    ruta_figura,
    dpi=300,
    bbox_inches="tight",
    facecolor="white"
)


# ------------------------------------------------------------
# 14. MOSTRAR EL GRÁFICO
# ------------------------------------------------------------

plt.show()


# ------------------------------------------------------------
# 15. MOSTRAR RESUMEN DE RESULTADOS
# ------------------------------------------------------------

comuna_mayor = datos_grafico.iloc[0]

comuna_menor = datos_grafico.iloc[-1]

mayor_cobertura = datos_grafico.loc[
    datos_grafico["cobertura_temporal_pct"].idxmax()
]

menor_cobertura = datos_grafico.loc[
    datos_grafico["cobertura_temporal_pct"].idxmin()
]


print("\n" + "=" * 70)
print("RESUMEN DE LA VISUALIZACIÓN")
print("=" * 70)

print(
    "\nComunas representadas:",
    len(datos_grafico)
)

print("\nMayor mediana documental:")
print(
    f"{comuna_mayor['comuna_oficial']}: "
    f"{comuna_mayor['mediana_horas']:.2f} horas"
)

print("\nMenor mediana documental:")
print(
    f"{comuna_menor['comuna_oficial']}: "
    f"{comuna_menor['mediana_horas']:.2f} horas"
)

print("\nMayor cobertura temporal:")
print(
    f"{mayor_cobertura['comuna_oficial']}: "
    f"{mayor_cobertura['cobertura_temporal_pct']:.2f}%"
)

print("\nMenor cobertura temporal:")
print(
    f"{menor_cobertura['comuna_oficial']}: "
    f"{menor_cobertura['cobertura_temporal_pct']:.2f}%"
)

print("\nFigura exportada:")
print(ruta_figura)

print("\n" + "=" * 70)
print("GRÁFICO 5.3 GENERADO CORRECTAMENTE")
print("=" * 70)
```

#### 6.4.2. Recurrencia causal documental

La participación porcentual de la descripción predominante permite caracterizar el grado de concentración relativa de una categoría documental dentro de cada comuna.

Asimismo, las descripciones predominantes representan categorías documentales registradas y no necesariamente mecanismos físicos de falla independientes.

#### 6.4.3. Duración documental

Las estadísticas temporales permiten comparar los intervalos documentales elegibles mediante la mediana y el percentil 75.

La comparación de ambas medidas aporta información complementaria sobre los valores centrales y superiores de las distribuciones registradas.

Por ello, la cobertura temporal constituye una dimensión necesaria para contextualizar las comparaciones entre comunas.

Las diferencias observadas en los intervalos documentales no permiten establecer directamente diferencias en las duraciones efectivas de interrupciones experimentadas por los usuarios.

### 6.5. Importancia de la cobertura y de las unidades de análisis

La cobertura temporal compara los registros elegibles con las asociaciones documentales disponibles.

Por esta razón, los porcentajes y tasas no deben combinarse directamente ni interpretarse como magnitudes equivalentes.

Una integración técnicamente rigurosa requiere conservar el significado de cada indicador, sus unidades y su población de referencia.

### 7.3. Limitaciones metodológicas

#### 7.3.1. Naturaleza documental de las fuentes

La principal limitación del proyecto se relaciona con la naturaleza de los registros utilizados.

Los informes del CEN constituyen antecedentes operacionales y administrativos del Sistema Eléctrico Nacional.

Por consiguiente, la presencia de un informe asociado con una comuna no demuestra necesariamente que todos sus habitantes hayan experimentado una interrupción eléctrica.

Asimismo, un informe documental no equivale necesariamente a un evento físico independiente.

Esta característica condiciona el alcance de los indicadores y exige que los resultados se interpreten como medidas de exposición operacional documentada, en lugar de estimaciones directas de interrupciones efectivas del suministro.

#### 7.3.2. Asociación territorial de los informes

Durante F2 se desarrollaron procedimientos de homologación territorial para vincular los registros documentales con las comunas seleccionadas.

Sin embargo, la información territorial puede presentar denominaciones heterogéneas, registros incompletos o referencias que requieren interpretación y normalización.

Adicionalmente, un mismo informe puede encontrarse asociado con más de una comuna.

Esta condición hace necesario distinguir entre:

- Informes documentales únicos.
- Asociaciones informe–comuna.
- Registros temporales elegibles.

La confusión entre estas unidades podría provocar duplicaciones conceptuales, interpretaciones incorrectas de los conteos o utilización de denominadores inadecuados.

La arquitectura desarrollada durante F3 permite organizar estas operaciones, pero la confiabilidad de las asociaciones continúa dependiendo de la información disponible y de las reglas de homologación aplicadas.

#### 7.3.3. Alcance territorial y temporal

La investigación se concentra en las 15 comunas más pobladas de Chile, seleccionadas mediante información del Censo 2024.

Por lo tanto, los resultados no representan necesariamente el comportamiento de todas las comunas del país.

El período documental analizado comprende 33 meses, entre enero de 2024 y septiembre de 2026.

Las comparaciones se encuentran delimitadas por dicho intervalo y por la disponibilidad de registros correspondientes a las fuentes utilizadas.

En consecuencia, los resultados no deben extrapolarse automáticamente a períodos posteriores ni utilizarse para establecer tendencias de largo plazo sin incorporar información adicional y verificar su comparabilidad.

#### 7.3.4. Normalización poblacional

La frecuencia documental se expresa por 100.000 habitantes-año, utilizando como referencia la población del Censo 2024.

Esta transformación permite comparar comunas con diferentes tamaños poblacionales.

Sin embargo, debe considerarse que la población constituye un denominador territorial y no una medida directa de usuarios afectados, clientes conectados, energía interrumpida o duración efectiva de los cortes.

Asimismo, se utiliza una población de referencia para el período completo de observación, por lo que el indicador no incorpora variaciones demográficas ocurridas durante los 33 meses.

En consecuencia, la normalización mejora la comparabilidad descriptiva entre comunas, pero no convierte la frecuencia documental en una probabilidad individual de interrupción eléctrica.

Tampoco permite reemplazar indicadores regulatorios de continuidad del suministro, como SAIFI o SAIDI.

#### 7.3.5. Clasificación de descripciones causales

El análisis de recurrencia utiliza las descripciones causales registradas en los documentos procesados.

Estas categorías constituyen denominaciones documentales y no necesariamente identificaciones definitivas de los mecanismos físicos responsables de una falla.

Además, se identificó la expresión «Balckout del sistema», conservada conforme a los valores presentes en el conjunto procesado.

Esta situación evidencia una oportunidad de mejora en la estandarización de denominaciones.

Una eventual normalización deberá preservar el valor original y documentar la equivalencia con la denominación estandarizada, evitando modificar silenciosamente los registros históricos.

Por otra parte, las visualizaciones de F4 representan únicamente la descripción predominante de cada comuna.

Esta selección facilita la comunicación de resultados, pero no permite examinar la distribución completa de las categorías causales de cada territorio.

#### 7.3.6. Cobertura y elegibilidad temporal

El análisis de duración se construyó a partir de intervalos documentales que cumplen las reglas de elegibilidad establecidas durante F2 y F3.

Las diferencias de cobertura muestran que no todas las asociaciones documentales disponen de información temporal apta para el cálculo de estadísticas.

Por ejemplo, Las Condes presenta una cobertura temporal del 23,68 %, mientras que La Florida alcanza el 60,23 %.

Esta heterogeneidad limita la comparación directa de las estadísticas temporales.

En particular, una mediana documental calculada sobre pocos registros elegibles requiere mayor cautela interpretativa que una estimación sustentada por una cantidad más amplia de observaciones.

Sin embargo, disponer de más registros no garantiza automáticamente que estos representen adecuadamente la totalidad de los eventos documentales.

Por ello, la interpretación debe considerar simultáneamente las estadísticas descriptivas, la cantidad de registros elegibles y los porcentajes de cobertura.

### 7.4. Riesgos de interpretación de los resultados

La comunicación de los resultados requiere controlar determinados riesgos que podrían conducir a conclusiones no respaldadas por la metodología.

**Riesgo de confundir frecuencia documental con interrupciones efectivas.** Una mayor frecuencia de informes no demuestra necesariamente que una comuna experimente más cortes eléctricos.

**Riesgo de interpretar la normalización como probabilidad individual.** La tasa por 100.000 habitantes-año constituye una medida descriptiva y no una estimación de riesgo individual.

**Riesgo de atribuir causalidad física a las descripciones documentales.** La recurrencia de una categoría no demuestra que represente el mecanismo causal de todos los registros asociados.

**Riesgo de interpretar intervalos documentales como duración efectiva de interrupciones.** Las estadísticas temporales corresponden a intervalos entre hitos registrados y no necesariamente a tiempos de afectación del suministro eléctrico.

**Riesgo de ignorar las diferencias de cobertura.** La interpretación de las medianas y percentiles sin considerar los registros elegibles puede producir comparaciones incompletas.

**Riesgo de combinar indicadores no equivalentes.** Las dimensiones analizadas utilizan diferentes unidades, denominadores y criterios de construcción, por lo que no resulta técnicamente adecuado integrarlas directamente en una puntuación territorial única sin desarrollar y justificar una metodología específica.

El reconocimiento de estos riesgos permite delimitar las conclusiones y reducir la posibilidad de atribuir a los indicadores significados que excedan su capacidad explicativa.

```python
# ============================================================
# MCDI500 - GRUPO 7
# FASE 4 - APARTADO 9.11
# EJECUCIÓN DE PRUEBAS AUTOMATIZADAS
# ============================================================

from pathlib import Path
from datetime import datetime
import importlib.util
import subprocess
import sys
import json
import os

print("=" * 65)
print("F4 - EJECUCIÓN DE PRUEBAS AUTOMATIZADAS")
print("=" * 65)


# ------------------------------------------------------------
# 1. IDENTIFICAR EL PROYECTO
# ------------------------------------------------------------

raiz = Path(raiz_proyecto).resolve()

directorio_pruebas = raiz / "tests"

directorio_resultados = raiz / "F4" / "resultados"

directorio_resultados.mkdir(
    parents=True,
    exist_ok=True
)


# ------------------------------------------------------------
# 2. IDENTIFICAR ARCHIVOS DE PRUEBA
# ------------------------------------------------------------

print("\n1. IDENTIFICACIÓN DE PRUEBAS")

archivos_prueba = []

if directorio_pruebas.is_dir():

    archivos_prueba = sorted(
        {
            *directorio_pruebas.rglob("test_*.py"),
            *directorio_pruebas.rglob("*_test.py")
        }
    )

print(
    "Directorio tests/:",
    "CORRECTO" if directorio_pruebas.is_dir()
    else "NO ENCONTRADO"
)

print(
    "Archivos de prueba identificados:",
    len(archivos_prueba)
)

for archivo in archivos_prueba:

    print(
        "-",
        archivo.relative_to(raiz)
    )


# ------------------------------------------------------------
# 3. COMPROBAR DISPONIBILIDAD DE PYTEST
# ------------------------------------------------------------

print("\n2. HERRAMIENTA DE PRUEBAS")

pytest_disponible = (
    importlib.util.find_spec("pytest") is not None
)

print(
    "pytest:",
    "DISPONIBLE" if pytest_disponible
    else "NO INSTALADO"
)


# ------------------------------------------------------------
# 4. EJECUTAR LAS PRUEBAS
# ------------------------------------------------------------

print("\n3. EJECUCIÓN")

estado = "NO EJECUTADO"
codigo_salida = None
salida_completa = ""

if not directorio_pruebas.is_dir():

    estado = "PENDIENTE: DIRECTORIO NO ENCONTRADO"

elif not pytest_disponible:

    estado = "PENDIENTE: PYTEST NO INSTALADO"

else:

    comando = [
        sys.executable,
        "-m",
        "pytest",
        "tests",
        "-q",
        "-p",
        "no:cacheprovider"
    ]

    entorno = os.environ.copy()

    # Evitar generar archivos __pycache__.
    entorno["PYTHONDONTWRITEBYTECODE"] = "1"

    # Permitir localizar los módulos del proyecto.
    entorno["PYTHONPATH"] = (
        str(raiz)
        + os.pathsep
        + entorno.get("PYTHONPATH", "")
    )

    print("Ejecutando pruebas del repositorio...")

    try:

        resultado = subprocess.run(
            comando,
            cwd=raiz,
            capture_output=True,
            text=True,
            encoding="utf-8",
            errors="replace",
            timeout=180,
            env=entorno
        )

        codigo_salida = resultado.returncode

        salida_completa = (
            resultado.stdout
            + "\n"
            + resultado.stderr
        )

        if codigo_salida == 0:

            estado = "PRUEBAS APROBADAS"

        elif codigo_salida == 5:

            estado = "NO SE RECOLECTARON PRUEBAS"

        else:

            estado = "PRUEBAS CON OBSERVACIONES"

    except subprocess.TimeoutExpired as error:

        estado = "TIEMPO DE EJECUCIÓN EXCEDIDO"

        salida_stdout = error.stdout or b""
        salida_stderr = error.stderr or b""

        if isinstance(salida_stdout, bytes):
            salida_stdout = salida_stdout.decode(
                "utf-8",
                errors="replace"
            )

        if isinstance(salida_stderr, bytes):
            salida_stderr = salida_stderr.decode(
                "utf-8",
                errors="replace"
            )

        salida_completa = (
            salida_stdout
            + "\n"
            + salida_stderr
        )


# ------------------------------------------------------------
# 5. MOSTRAR LOS RESULTADOS
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("RESULTADOS DE LAS PRUEBAS")
print("=" * 65)

if salida_completa.strip():

    lineas = salida_completa.strip().splitlines()

    # Mostrar las últimas líneas para evitar
    # una salida excesivamente extensa.

    for linea in lineas[-30:]:
        print(linea)

else:

    print(
        "No se dispone de resultados de ejecución."
    )


# ------------------------------------------------------------
# 6. GUARDAR EVIDENCIA
# ------------------------------------------------------------

ruta_salida = (
    directorio_resultados
    / "F4_902_Salida_Pruebas.txt"
)

ruta_json = (
    directorio_resultados
    / "F4_902_Pruebas_Automatizadas.json"
)

ruta_salida.write_text(
    salida_completa,
    encoding="utf-8"
)

registro = {
    "proyecto": "MCDI500 - Grupo 7",
    "fase": "F4",
    "apartado": "9.11",
    "fecha": datetime.now().isoformat(
        timespec="seconds"
    ),
    "interprete": sys.executable,
    "directorio_pruebas": str(directorio_pruebas),
    "archivos_identificados": [
        str(archivo.relative_to(raiz))
        for archivo in archivos_prueba
    ],
    "pytest_disponible": pytest_disponible,
    "estado": estado,
    "codigo_salida": codigo_salida,
    "archivo_resultados": str(ruta_salida)
}

with open(
    ruta_json,
    "w",
    encoding="utf-8"
) as archivo:

    json.dump(
        registro,
        archivo,
        ensure_ascii=False,
        indent=4
    )


# ------------------------------------------------------------
# 7. RESUMEN FINAL
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("RESUMEN DE VERIFICACIÓN 9.11")
print("=" * 65)

print(
    "Archivos de prueba identificados:",
    len(archivos_prueba)
)

print(
    "pytest disponible:",
    "SÍ" if pytest_disponible else "NO"
)

print("Estado:", estado)

print("Código de salida:", codigo_salida)

print("\nEvidencias guardadas en:")
print(ruta_salida)
print(ruta_json)

print("=" * 65)
```

```python
# ============================================================
# MCDI500 - GRUPO 7
# FASE 4 - APARTADO 9.13
# VERIFICACIÓN DE DEPENDENCIAS Y DOCUMENTACIÓN
# ============================================================

from pathlib import Path
from datetime import datetime
from importlib import metadata
import subprocess
import sys
import json

print("=" * 65)
print("F4 - VERIFICACIÓN DE DEPENDENCIAS Y DOCUMENTACIÓN")
print("=" * 65)


# ------------------------------------------------------------
# 1. IDENTIFICAR ARCHIVOS
# ------------------------------------------------------------

raiz = Path(raiz_proyecto).resolve()

ruta_requirements = raiz / "requirements.txt"
ruta_readme = raiz / "README.md"

directorio_resultados = raiz / "F4" / "resultados"

directorio_resultados.mkdir(
    parents=True,
    exist_ok=True
)


# ------------------------------------------------------------
# 2. COMPROBAR DEPENDENCIAS FUNDAMENTALES
# ------------------------------------------------------------

print("\n1. BIBLIOTECAS INSTALADAS")

bibliotecas = [
    "pandas",
    "numpy",
    "matplotlib",
    "pytest"
]

versiones = {}

for nombre in bibliotecas:

    try:

        version = metadata.version(nombre)

        versiones[nombre] = version

        print(f"{nombre}: {version}")

    except metadata.PackageNotFoundError:

        versiones[nombre] = None

        print(f"{nombre}: NO INSTALADA")


# ------------------------------------------------------------
# 3. ANALIZAR REQUIREMENTS.TXT
# ------------------------------------------------------------

print("\n2. DEPENDENCIAS DECLARADAS EN REQUIREMENTS.TXT")

requirements_existe = ruta_requirements.is_file()

dependencias_evaluadas = []
dependencias_no_evaluadas = []

if requirements_existe:

    from packaging.requirements import Requirement

    lineas = ruta_requirements.read_text(
        encoding="utf-8-sig"
    ).splitlines()

    for numero, linea in enumerate(lineas, start=1):

        texto = linea.strip()

        if not texto or texto.startswith("#"):
            continue

        # Se excluyen comentarios al final de línea.
        texto = texto.split(" #", 1)[0].strip()

        try:

            requisito = Requirement(texto)

            # Respetar condiciones de entorno.
            if (
                requisito.marker is not None
                and not requisito.marker.evaluate()
            ):
                continue

            nombre = requisito.name

            try:

                version_instalada = metadata.version(nombre)

                cumple_version = (
                    requisito.specifier.contains(
                        version_instalada,
                        prereleases=True
                    )
                    if requisito.specifier
                    else True
                )

                estado = (
                    "CORRECTO"
                    if cumple_version
                    else "VERSIÓN NO COMPATIBLE"
                )

            except metadata.PackageNotFoundError:

                version_instalada = None
                estado = "NO INSTALADA"

            dependencias_evaluadas.append({
                "nombre": nombre,
                "version_instalada": version_instalada,
                "estado": estado
            })

            print(f"{nombre}: {estado}")

        except Exception:

            # Ejemplo: referencias a otros archivos,
            # opciones de pip o dependencias especiales.
            # No se interpretan automáticamente.

            dependencias_no_evaluadas.append(numero)

            print(
                f"Línea {numero}: "
                "REQUIERE REVISIÓN MANUAL"
            )

else:

    print("requirements.txt: NO ENCONTRADO")


# ------------------------------------------------------------
# 4. COMPROBAR DECLARACIÓN DE PYTEST
# ------------------------------------------------------------

print("\n3. DECLARACIÓN DE PYTEST")

pytest_declarado = any(
    dependencia["nombre"].lower() == "pytest"
    for dependencia in dependencias_evaluadas
)

print(
    "pytest declarado directamente:",
    "SÍ" if pytest_declarado else "NO"
)

if not pytest_declarado:

    print(
        "OBSERVACIÓN: revisar si pytest se declara "
        "en otro archivo de dependencias."
    )


# ------------------------------------------------------------
# 5. REVISAR DOCUMENTACIÓN README
# ------------------------------------------------------------

print("\n4. DOCUMENTACIÓN README")

readme_existe = ruta_readme.is_file()

if readme_existe:

    contenido_readme = ruta_readme.read_text(
        encoding="utf-8-sig",
        errors="replace"
    ).lower()

    menciona_dependencias = (
        "requirements.txt" in contenido_readme
        or "pip install" in contenido_readme
    )

    menciona_pruebas = (
        "pytest" in contenido_readme
    )

    print("README.md: CORRECTO")

    print(
        "Menciona instalación de dependencias:",
        "SÍ" if menciona_dependencias else "NO"
    )

    print(
        "Menciona ejecución de pytest:",
        "SÍ" if menciona_pruebas else "NO"
    )

else:

    menciona_dependencias = False
    menciona_pruebas = False

    print("README.md: NO ENCONTRADO")


# ------------------------------------------------------------
# 6. VERIFICAR CONSISTENCIA DEL ENTORNO
# ------------------------------------------------------------

print("\n5. CONSISTENCIA DE DEPENDENCIAS INSTALADAS")

try:

    resultado_pip = subprocess.run(
        [
            sys.executable,
            "-m",
            "pip",
            "check"
        ],
        cwd=raiz,
        capture_output=True,
        text=True,
        encoding="utf-8",
        errors="replace",
        timeout=60
    )

    pip_check_correcto = (
        resultado_pip.returncode == 0
    )

    salida_pip = (
        resultado_pip.stdout
        + "\n"
        + resultado_pip.stderr
    ).strip()

    print(salida_pip)

except Exception as error:

    pip_check_correcto = False

    salida_pip = str(error)

    print("No fue posible completar pip check.")
    print(salida_pip)


# ------------------------------------------------------------
# 7. IDENTIFICAR OBSERVACIONES
# ------------------------------------------------------------

dependencias_con_observaciones = [
    dependencia
    for dependencia in dependencias_evaluadas
    if dependencia["estado"] != "CORRECTO"
]


# ------------------------------------------------------------
# 8. GUARDAR EVIDENCIA
# ------------------------------------------------------------

registro = {
    "proyecto": "MCDI500 - Grupo 7",
    "fase": "F4",
    "apartado": "9.13",
    "fecha": datetime.now().isoformat(
        timespec="seconds"
    ),
    "interprete": sys.executable,
    "versiones_bibliotecas": versiones,
    "requirements_existe": requirements_existe,
    "dependencias_evaluadas": dependencias_evaluadas,
    "lineas_no_evaluadas": dependencias_no_evaluadas,
    "pytest_declarado_directamente": pytest_declarado,
    "readme_existe": readme_existe,
    "readme_menciona_dependencias": menciona_dependencias,
    "readme_menciona_pruebas": menciona_pruebas,
    "pip_check_correcto": pip_check_correcto,
    "salida_pip_check": salida_pip
}

ruta_registro = (
    directorio_resultados
    / "F4_903_Verificacion_Dependencias.json"
)

with open(
    ruta_registro,
    "w",
    encoding="utf-8"
) as archivo:

    json.dump(
        registro,
        archivo,
        ensure_ascii=False,
        indent=4
    )


# ------------------------------------------------------------
# 9. MOSTRAR RESUMEN FINAL
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("RESUMEN DE VERIFICACIÓN 9.13")
print("=" * 65)

print(
    "requirements.txt:",
    "DISPONIBLE" if requirements_existe
    else "NO DISPONIBLE"
)

print(
    "Dependencias evaluadas:",
    len(dependencias_evaluadas)
)

print(
    "Dependencias con observaciones:",
    len(dependencias_con_observaciones)
)

print(
    "Líneas que requieren revisión manual:",
    len(dependencias_no_evaluadas)
)

print(
    "pytest declarado directamente:",
    "SÍ" if pytest_declarado else "NO"
)

print(
    "README.md:",
    "DISPONIBLE" if readme_existe
    else "NO DISPONIBLE"
)

print(
    "README menciona instalación:",
    "SÍ" if menciona_dependencias else "NO"
)

print(
    "README menciona pytest:",
    "SÍ" if menciona_pruebas else "NO"
)

print(
    "Consistencia del entorno (pip check):",
    "CORRECTO" if pip_check_correcto
    else "REVISAR"
)

print("\nEvidencia guardada:")
print(ruta_registro)

print("=" * 65)
```

```python
# ============================================================
# MCDI500 - GRUPO 7
# FASE 4 - APARTADO 9.15
# AUDITORÍA DE INTEGRIDAD Y TRAZABILIDAD
# ============================================================

from pathlib import Path
from datetime import datetime
import pandas as pd
import hashlib
import json
import struct


print("=" * 65)
print("F4 - AUDITORÍA DE PRODUCTOS Y TRAZABILIDAD")
print("=" * 65)


# ------------------------------------------------------------
# 1. IDENTIFICAR ARCHIVOS
# ------------------------------------------------------------

raiz = Path(raiz_proyecto).resolve()

ruta_panel = (
    raiz
    / "F3"
    / "resultados"
    / "F3_701_Panel_Integrado_Comunal.csv"
)

directorio_tablas = raiz / "F4" / "resultados"
directorio_figuras = raiz / "F4" / "figuras"

if not ruta_panel.is_file():
    raise FileNotFoundError(
        f"No se encontró el panel F3: {ruta_panel}"
    )

panel = pd.read_csv(ruta_panel)

print("\n1. ARCHIVO DE ORIGEN")
print("Panel F3:", ruta_panel.name)
print("Comunas:", len(panel))
print("Variables:", len(panel.columns))


# ------------------------------------------------------------
# 2. DEFINIR LOS PRODUCTOS QUE DEBEN VERIFICARSE
# ------------------------------------------------------------

configuracion_tablas = {
    "F4_Tabla_401_Frecuencia_Comunal.csv": [
        "comuna_oficial",
        "poblacion_2024",
        "informes_no_anulados",
        "frecuencia_100k_hab_anio"
    ],

    "F4_Tabla_402_Recurrencia_Causal.csv": [
        "comuna_oficial",
        "asociaciones_documentales",
        "causa_mas_recurrente",
        "informes_causa_principal",
        "porcentaje_causa_principal"
    ],

    "F4_Tabla_403_Duracion_Cobertura.csv": [
        "comuna_oficial",
        "informes_validos",
        "mediana_horas",
        "percentil_75_horas",
        "cobertura_temporal_pct"
    ]
}

archivos_figuras = [
    "F4_Figura_501_Frecuencia_Comunal.png",
    "F4_Figura_502_Recurrencia_Causal.png",
    "F4_Figura_503_Duracion_Cobertura.png"
]


# ------------------------------------------------------------
# 3. FUNCIÓN PARA CALCULAR SHA-256
# ------------------------------------------------------------

def calcular_sha256(ruta):

    calculador = hashlib.sha256()

    with open(ruta, "rb") as archivo:

        for bloque in iter(
            lambda: archivo.read(1024 * 1024),
            b""
        ):
            calculador.update(bloque)

    return calculador.hexdigest()


# ------------------------------------------------------------
# 4. VERIFICAR LAS TRES TABLAS
# ------------------------------------------------------------

print("\n2. CONSISTENCIA DE TABLAS F4 CON PANEL F3")

resultados_tablas = {}

for nombre, columnas in configuracion_tablas.items():

    ruta_tabla = directorio_tablas / nombre

    resultado = {
        "archivo": nombre,
        "existe": ruta_tabla.is_file(),
        "coincide_con_panel_f3": False,
        "filas": None,
        "sha256": None,
        "observacion": None
    }

    if ruta_tabla.is_file():

        try:

            tabla = pd.read_csv(ruta_tabla)

            resultado["filas"] = len(tabla)

            # Recuperar exactamente las mismas variables
            # desde el panel original de F3.

            esperado = panel[columnas].copy()

            obtenido = tabla[columnas].copy()

            # Ordenar por comuna para evitar que el
            # orden visual afecte la comparación.

            esperado = esperado.sort_values(
                "comuna_oficial"
            ).reset_index(drop=True)

            obtenido = obtenido.sort_values(
                "comuna_oficial"
            ).reset_index(drop=True)

            # Verificar identidad de las 15 comunas.

            if len(tabla) != 15:
                raise ValueError(
                    "La tabla no contiene 15 registros."
                )

            if tabla["comuna_oficial"].nunique() != 15:
                raise ValueError(
                    "Existen comunas duplicadas."
                )

            # Comparar todas las variables.
            # Se permite únicamente tolerancia numérica
            # mínima por lectura y escritura de decimales.

            pd.testing.assert_frame_equal(
                esperado,
                obtenido,
                check_dtype=False,
                check_exact=False,
                rtol=1e-9,
                atol=1e-9
            )

            resultado["coincide_con_panel_f3"] = True

            resultado["sha256"] = calcular_sha256(
                ruta_tabla
            )

            resultado["observacion"] = (
                "Contenido consistente con F3"
            )

            print(
                f"✓ {nombre}: "
                "15 comunas - CONSISTENTE"
            )

        except Exception as error:

            resultado["observacion"] = str(error)

            print(
                f"✗ {nombre}: REVISAR"
            )

            print("  Motivo:", error)

    else:

        resultado["observacion"] = (
            "Archivo no encontrado"
        )

        print(
            f"✗ {nombre}: NO ENCONTRADO"
        )

    resultados_tablas[nombre] = resultado


# ------------------------------------------------------------
# 5. VERIFICAR LAS TRES FIGURAS
# ------------------------------------------------------------

print("\n3. INTEGRIDAD DE FIGURAS PNG")

resultados_figuras = {}

firma_png = b"\x89PNG\r\n\x1a\n"

for nombre in archivos_figuras:

    ruta_figura = directorio_figuras / nombre

    resultado = {
        "archivo": nombre,
        "existe": ruta_figura.is_file(),
        "formato_png_valido": False,
        "ancho_pixeles": None,
        "alto_pixeles": None,
        "tamano_bytes": None,
        "sha256": None
    }

    if ruta_figura.is_file():

        with open(ruta_figura, "rb") as archivo:
            encabezado = archivo.read(24)

        tamano = ruta_figura.stat().st_size

        resultado["tamano_bytes"] = tamano

        # Revisar firma y encabezado básico PNG.
        if (
            encabezado[:8] == firma_png
            and encabezado[12:16] == b"IHDR"
            and len(encabezado) == 24
        ):

            ancho, alto = struct.unpack(
                ">II",
                encabezado[16:24]
            )

            if ancho > 0 and alto > 0 and tamano > 1000:

                resultado["formato_png_valido"] = True

                resultado["ancho_pixeles"] = ancho
                resultado["alto_pixeles"] = alto

                resultado["sha256"] = calcular_sha256(
                    ruta_figura
                )

                print(
                    f"✓ {nombre}: "
                    f"{ancho} x {alto} píxeles"
                )

        if not resultado["formato_png_valido"]:

            print(
                f"✗ {nombre}: REVISAR"
            )

    else:

        print(
            f"✗ {nombre}: NO ENCONTRADO"
        )

    resultados_figuras[nombre] = resultado


# ------------------------------------------------------------
# 6. REGISTRAR TRAZABILIDAD DEL PANEL ORIGINAL
# ------------------------------------------------------------

print("\n4. TRAZABILIDAD DEL ARCHIVO DE ORIGEN")

sha_panel = calcular_sha256(ruta_panel)

print("Archivo:", ruta_panel.name)
print("SHA-256:", sha_panel)


# ------------------------------------------------------------
# 7. RESUMEN FINAL
# ------------------------------------------------------------

tablas_correctas = sum(
    resultado["coincide_con_panel_f3"]
    for resultado in resultados_tablas.values()
)

figuras_correctas = sum(
    resultado["formato_png_valido"]
    for resultado in resultados_figuras.values()
)

print("\n" + "=" * 65)
print("RESUMEN DE AUDITORÍA 9.15")
print("=" * 65)

print(
    "Tablas consistentes con F3:",
    f"{tablas_correctas}/3"
)

print(
    "Figuras con encabezado PNG válido:",
    f"{figuras_correctas}/3"
)

auditoria_satisfactoria = (
    tablas_correctas == 3
    and figuras_correctas == 3
)

print(
    "Resultado general:",
    "CONFORME"
    if auditoria_satisfactoria
    else "CON OBSERVACIONES"
)


# ------------------------------------------------------------
# 8. GUARDAR EVIDENCIA DE AUDITORÍA
# ------------------------------------------------------------

registro = {
    "proyecto": "MCDI500 - Grupo 7",
    "fase": "F4",
    "apartado": "9.15",
    "fecha": datetime.now().isoformat(
        timespec="seconds"
    ),
    "panel_origen": str(ruta_panel),
    "sha256_panel_f3": sha_panel,
    "tablas": resultados_tablas,
    "figuras": resultados_figuras,
    "tablas_correctas": tablas_correctas,
    "figuras_correctas": figuras_correctas,
    "resultado_general": (
        "CONFORME"
        if auditoria_satisfactoria
        else "CON OBSERVACIONES"
    )
}

ruta_registro = (
    directorio_tablas
    / "F4_904_Auditoria_Productos.json"
)

with open(
    ruta_registro,
    "w",
    encoding="utf-8"
) as archivo:

    json.dump(
        registro,
        archivo,
        ensure_ascii=False,
        indent=4
    )

print("\nEvidencia guardada:")
print(ruta_registro)

print("=" * 65)
```

```python
# ============================================================
# MCDI500 - GRUPO 7
# FASE 4 - APARTADO 9.17
# VERIFICACIÓN DE TRAZABILIDAD Y CONTROL DE VERSIONES
# ============================================================

from pathlib import Path
from datetime import datetime
import subprocess
import shutil
import json

print("=" * 65)
print("F4 - TRAZABILIDAD Y CONTROL DE VERSIONES")
print("=" * 65)


# ------------------------------------------------------------
# 1. IDENTIFICAR EL REPOSITORIO
# ------------------------------------------------------------

raiz = Path(raiz_proyecto).resolve()

directorio_resultados = raiz / "F4" / "resultados"

directorio_resultados.mkdir(
    parents=True,
    exist_ok=True
)


# ------------------------------------------------------------
# 2. FUNCIÓN PARA CONSULTAR GIT
# ------------------------------------------------------------

def consultar_git(argumentos):

    try:

        resultado = subprocess.run(
            ["git"] + argumentos,
            cwd=raiz,
            capture_output=True,
            text=True,
            encoding="utf-8",
            errors="replace",
            timeout=30
        )

        return {
            "correcto": resultado.returncode == 0,
            "salida": resultado.stdout.strip(),
            "error": resultado.stderr.strip(),
            "codigo": resultado.returncode
        }

    except Exception as error:

        return {
            "correcto": False,
            "salida": "",
            "error": str(error),
            "codigo": None
        }


# ------------------------------------------------------------
# 3. VERIFICAR DISPONIBILIDAD DE GIT
# ------------------------------------------------------------

print("\n1. DISPONIBILIDAD DE GIT")

git_disponible = shutil.which("git") is not None

print(
    "Git instalado:",
    "SÍ" if git_disponible else "NO"
)

if not git_disponible:

    print(
        "No es posible completar la auditoría "
        "sin Git."
    )


# ------------------------------------------------------------
# 4. IDENTIFICAR RAMA Y COMMIT
# ------------------------------------------------------------

print("\n2. IDENTIFICACIÓN DEL REPOSITORIO")

consulta_repositorio = consultar_git(
    ["rev-parse", "--is-inside-work-tree"]
)

repositorio_valido = (
    consulta_repositorio["correcto"]
    and consulta_repositorio["salida"] == "true"
)

if repositorio_valido:

    consulta_rama = consultar_git(
        ["branch", "--show-current"]
    )

    consulta_commit = consultar_git(
        ["rev-parse", "--short", "HEAD"]
    )

    consulta_raiz = consultar_git(
        ["rev-parse", "--show-toplevel"]
    )

    rama = consulta_rama["salida"]
    commit = consulta_commit["salida"]

    print("Repositorio Git: CORRECTO")
    print("Rama actual:", rama)
    print("Último commit:", commit)

else:

    rama = None
    commit = None

    print("Repositorio Git: REVISAR")


# ------------------------------------------------------------
# 5. IDENTIFICAR CAMBIOS PENDIENTES
# ------------------------------------------------------------

print("\n3. CAMBIOS PENDIENTES DE INTEGRACIÓN")

cambios = []

if repositorio_valido:

    consulta_estado = consultar_git(
        [
            "status",
            "--short",
            "--untracked-files=normal"
        ]
    )

    if consulta_estado["correcto"]:

        cambios = [
            linea
            for linea in consulta_estado["salida"].splitlines()
            if linea.strip()
        ]

        print(
            "Entradas con cambios pendientes:",
            len(cambios)
        )

        if cambios:

            for cambio in cambios[:20]:
                print("-", cambio)

            if len(cambios) > 20:

                print(
                    f"... y {len(cambios) - 20} "
                    "entradas adicionales."
                )

        else:

            print(
                "No se identificaron cambios pendientes."
            )

    else:

        print("No fue posible consultar git status.")


# ------------------------------------------------------------
# 6. REVISAR ARCHIVOS DOCUMENTALES
# ------------------------------------------------------------

print("\n4. ARCHIVOS DE DOCUMENTACIÓN")

archivos_documentacion = [
    "README.md",
    "requirements.txt",
    "CHANGELOG.md",
    "changelog.md",
    "docs/changelog.md"
]

estado_documentacion = {}

for nombre in archivos_documentacion:

    disponible = (
        raiz / nombre
    ).is_file()

    estado_documentacion[nombre] = disponible

    print(
        f"{nombre}:",
        "DISPONIBLE" if disponible else "NO ENCONTRADO"
    )


# ------------------------------------------------------------
# 7. CONSULTAR HISTORIAL RECIENTE
# ------------------------------------------------------------

print("\n5. HISTORIAL RECIENTE DE COMMITS")

historial = []

if repositorio_valido:

    consulta_historial = consultar_git(
        [
            "log",
            "-5",
            "--date=short",
            "--pretty=format:%h | %ad | %an | %s"
        ]
    )

    if consulta_historial["correcto"]:

        historial = (
            consulta_historial["salida"].splitlines()
        )

        for linea in historial:
            print("-", linea)

    else:

        print("No fue posible consultar el historial.")


# ------------------------------------------------------------
# 8. VERIFICAR ARCHIVOS PRINCIPALES DE F4
# ------------------------------------------------------------

print("\n6. PRODUCTOS PRINCIPALES DE F4")

productos_f4 = [
    "F4/F4_Comunicacion_Resultados.ipynb",
    "F4/resultados/F4_Tabla_401_Frecuencia_Comunal.csv",
    "F4/resultados/F4_Tabla_402_Recurrencia_Causal.csv",
    "F4/resultados/F4_Tabla_403_Duracion_Cobertura.csv",
    "F4/figuras/F4_Figura_501_Frecuencia_Comunal.png",
    "F4/figuras/F4_Figura_502_Recurrencia_Causal.png",
    "F4/figuras/F4_Figura_503_Duracion_Cobertura.png"
]

estado_productos = {}

for nombre in productos_f4:

    disponible = (
        raiz / nombre
    ).is_file()

    estado_productos[nombre] = disponible

    print(
        f"{nombre}:",
        "CORRECTO" if disponible else "REVISAR"
    )


# ------------------------------------------------------------
# 9. GUARDAR EVIDENCIA
# ------------------------------------------------------------

registro = {
    "proyecto": "MCDI500 - Grupo 7",
    "fase": "F4",
    "apartado": "9.17",
    "fecha": datetime.now().isoformat(
        timespec="seconds"
    ),
    "git_disponible": git_disponible,
    "repositorio_valido": repositorio_valido,
    "rama_actual": rama,
    "ultimo_commit": commit,
    "cambios_pendientes": cambios,
    "documentacion": estado_documentacion,
    "historial_reciente": historial,
    "productos_f4": estado_productos
}

ruta_registro = (
    directorio_resultados
    / "F4_905_Trazabilidad_Git.json"
)

with open(
    ruta_registro,
    "w",
    encoding="utf-8"
) as archivo:

    json.dump(
        registro,
        archivo,
        ensure_ascii=False,
        indent=4
    )


# ------------------------------------------------------------
# 10. RESUMEN FINAL
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("RESUMEN DE VERIFICACIÓN 9.17")
print("=" * 65)

print(
    "Repositorio Git:",
    "CORRECTO" if repositorio_valido else "REVISAR"
)

print("Rama actual:", rama)

print("Último commit:", commit)

print(
    "Entradas con cambios pendientes:",
    len(cambios)
)

print(
    "Productos F4 disponibles:",
    f"{sum(estado_productos.values())}/7"
)

print(
    "Archivo changelog disponible:",
    any(
        estado_documentacion[nombre]
        for nombre in [
            "CHANGELOG.md",
            "changelog.md",
            "docs/changelog.md"
        ]
    )
)

print("\nEvidencia guardada:")
print(ruta_registro)

print("=" * 65)
```
